```python
import requests
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
url = "https://api.openalex.org/works?search=artificial%20intelligence&per-page=100"

response = requests.get(url)

data = response.json()

print("Total Papers:", len(data["results"]))
```

```python
papers = []

for work in data["results"]:
    papers.append({
            "Title": work.get("title"),
                    "Publication_Year": work.get("publication_year"),
                            "Citations": work.get("cited_by_count"),
                                    "Publication_Type": work.get("type"),
                                            "DOI": work.get("doi"),
                                                    "OpenAlex_ID": work.get("id")
                                                        })

df = pd.DataFrame(papers)
df.head()
```

```python
df.to_csv("OpenAlex_AI_Research.csv", index=False)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
df["Publication_Year"].value_counts().sort_index().plot(kind="line", marker="o")

plt.title("AI Research Publications Over Years")
plt.xlabel("Publication Year")
plt.ylabel("Number of Publications")
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(8,5))
df["Publication_Type"].value_counts().plot(kind="bar")

plt.title("Distribution of Publication Types")
plt.xlabel("Publication Type")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Citations"], bins=20)

plt.title("Citation Distribution")
plt.xlabel("Citation Count")
plt.ylabel("Frequency")

plt.show()
```

```python
top10 = df.sort_values(by="Citations", ascending=False).head(10)

top10[["Title","Citations"]]
```

```python
plt.figure(figsize=(10,6))

plt.barh(top10["Title"], top10["Citations"])

plt.title("Top 10 Most Cited AI Papers")

plt.xlabel("Citation Count")

plt.show()
```
