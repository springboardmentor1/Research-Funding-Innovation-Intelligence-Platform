```python
papers = papers.drop_duplicates()
grants = grants.drop_duplicates()
patents = patents.drop_duplicates()
```

```python
for df in [papers, grants, patents]:
    text_columns = df.select_dtypes(include="object").columns
    df[text_columns] = df[text_columns].fillna("Unknown")
```

```python
papers["published"] = pd.to_datetime(papers["published"])
```

```python
papers["publication_year"] = papers["published"].dt.year
```

```python
import matplotlib.pyplot as plt

papers.isnull().sum().sort_values().plot(kind="bar", figsize=(10,5))
plt.title("Missing Values in Papers Dataset")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
papers["primary_category"].value_counts().head(10).plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Top 10 Research Categories")
plt.xlabel("Category")
plt.ylabel("Number of Papers")
plt.xticks(rotation=45)
plt.show()
```
