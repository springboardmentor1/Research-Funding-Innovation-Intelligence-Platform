```python
papers = papers.drop_duplicates()
grants = grants.drop_duplicates()
patents = patents.drop_duplicates()
```

```python
papers.fillna("Unknown", inplace=True)
grants.fillna("Unknown", inplace=True)
patents.fillna("Unknown", inplace=True)
```

```python
papers["published"] = pd.to_datetime(papers["published"])
```

```python
papers["publication_year"] = papers["published"].dt.year
```
