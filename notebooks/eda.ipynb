```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# load the clean CSV files
df_pubs = pd.read_csv("../data/processed/publications_clean.csv")
df_grants = pd.read_csv("../data/processed/grants_clean.csv")
df_patents = pd.read_csv("../data/processed/patents_clean.csv")

print(df_pubs.head())
print(df_grants.head())
print(df_patents.head())
```

```python
# count papers per year and plot
year_counts = df_pubs["year"].value_counts().sort_index()
plt.plot(year_counts.index, year_counts.values)
plt.title("Publications per year")
plt.show()
```

```python
# top domains by publication count
top_domains = df_pubs["domain"].value_counts().head(10)
top_domains.plot(kind="bar")
plt.title("Top domains")
plt.show()
```

```python
# distribution of citation counts
df_pubs["cited_by_count"].hist(bins=20)
plt.title("Citations distribution")
plt.show()
```

```python
# top funders by award counts
top_funders = df_grants["funder_name"].value_counts().head(10)
top_funders.plot(kind="bar")
plt.title("Top funders")
plt.show()
```

```python
# distribution of linked papers per grant
df_grants["linked_works_count"].hist(bins=15)
plt.title("Linked works per award")
plt.show()
```

```python
# patents filed per year
df_patents["filing_year"] = pd.to_datetime(df_patents["filing_date"]).dt.year
patent_years = df_patents["filing_year"].value_counts().sort_index()
patent_years.plot(kind="bar")
plt.title("Patents per year")
plt.show()
```

```python
# top patent assignees
top_assignees = df_patents["assignee"].value_counts().head(10)
top_assignees.plot(kind="bar")
plt.title("Top patent assignees")
plt.show()
```

```python
# patents by technology domain code
pat_domains = df_patents["technology_domain"].value_counts().head(10)
pat_domains.plot(kind="bar")
plt.title("Patents by CPC class")
plt.show()
```

```python
# overlap check of title keywords
pub_titles = " ".join(df_pubs["title"].astype(str)).lower().split()
pat_titles = " ".join(df_patents["title"].astype(str)).lower().split()

pub_words = set(pub_titles)
pat_words = set(pat_titles)

overlap = pub_words.intersection(pat_words)
print("Overlapping words in paper and patent titles:", list(overlap)[:20])
```
